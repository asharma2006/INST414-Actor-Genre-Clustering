```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import normalize
```

```python

import glob

files = glob.glob("imdb_movies_2000to2022.actorXgenre*.csv")

if not files:
    raise FileNotFoundError("Put the IMDb CSV file in the same folder as this notebook.")

df = pd.read_csv(files[0], index_col="actor_id")

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

print("\nColumns:")
print(df.columns.tolist())
```

```python

print("Missing values:", df.isnull().sum().sum())
print("Duplicate actors:", df.index.duplicated().sum())


df = df.drop(columns=["Unnamed: 22"])


# Keep numerical genre features
X = df.select_dtypes(include=[np.number]).copy()

# Remove features that never vary
X = X.loc[:, X.nunique() > 1]

# Fill missing values with zero
X = X.fillna(0)

# Remove rows without genre information
X = X.loc[X.sum(axis=1) > 0]

print("Cleaned dataset shape:", X.shape)
display(X.head())
```

```python

X_scaled = normalize(X, norm="l1")

print("Number of actors:", X_scaled.shape[0])
print("Number of genre features:", X_scaled.shape[1])
```

```python

k_values = range(2, 11)

inertias = []
silhouette_scores = []

for k in k_values:
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    inertias.append(model.inertia_)

    score = silhouette_score(
        X_scaled,
        labels,
        sample_size=min(2000, len(X_scaled)),
        random_state=42
    )

    silhouette_scores.append(score)

results = pd.DataFrame({
    "k": list(k_values),
    "Inertia": inertias,
    "Silhouette Score": silhouette_scores
})

display(results)
```

```python

plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    inertias,
    marker="o"
)

plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method for K-Means Clustering")
plt.xticks(list(k_values))
plt.grid(True)

plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Scores for Different k Values")
plt.xticks(list(k_values))
plt.grid(True)

plt.show()

best_k = int(results.loc[
    results["Silhouette Score"].idxmax(), "k"
])

print("Best k by silhouette score:", best_k)
```

```python

chosen_k = 5

kmeans = KMeans(n_clusters=chosen_k, random_state=42, n_init=10)

cluster_labels = kmeans.fit_predict(X_scaled)

clustered_df = X.copy()
clustered_df["Cluster"] = cluster_labels

print("Number of clusters:", chosen_k)
print("\nActors per cluster:")
print(clustered_df["Cluster"].value_counts().sort_index())

```

```python

cluster_sizes = (
    clustered_df["Cluster"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(8, 5))
cluster_sizes.plot(kind="bar")

plt.xlabel("Cluster")
plt.ylabel("Number of Actors")
plt.title("Number of Actors in Each Cluster")
plt.xticks(rotation=0)
plt.show()
```

```python

genre_profiles = pd.DataFrame(
    kmeans.cluster_centers_,
    columns=X.columns
)

for cluster in range(chosen_k):
    print(f"\nCLUSTER {cluster}")
    print("Top 5 genres:")

    print(
        genre_profiles.loc[cluster]
        .sort_values(ascending=False)
        .head(5)
    )
```

```python

for cluster in range(chosen_k):
    print(f"\nCLUSTER {cluster} EXAMPLES")

    members = clustered_df[
        clustered_df["Cluster"] == cluster
    ]

    member_positions = np.where(
        cluster_labels == cluster
    )[0]

    distances = np.linalg.norm(
        X_scaled[member_positions] -
        kmeans.cluster_centers_[cluster],
        axis=1
    )

    closest = np.argsort(distances)[:2]

    display(members.iloc[closest])
```

```python

top_genres = (
    genre_profiles.mean()
    .sort_values(ascending=False)
    .head(8)
    .index
)

genre_profiles[top_genres].T.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Genre Profiles Across Clusters")
plt.xlabel("Genre")
plt.ylabel("Average Normalized Genre Count")
plt.legend(title="Cluster")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

print("Validation Checks")

print("Missing values:", X.isnull().sum().sum())
print("Actors clustered:", len(cluster_labels))
print("Unique clusters:", len(np.unique(cluster_labels)))

print(
    "All actors assigned:",
    len(cluster_labels) == len(X)
)

print(
    "Recalculated silhouette score:",
    silhouette_score(
        X_scaled,
        cluster_labels,
        sample_size=min(2000, len(X_scaled)),
        random_state=42
    )
)

clustered_df.to_csv("actor_clusters_results.csv")
genre_profiles.to_csv("cluster_genre_profiles.csv")

print("\nResults saved successfully!")
```
